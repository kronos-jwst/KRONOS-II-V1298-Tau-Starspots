```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import corner
```

```python
## -- visit 1 samples
path1 = '/home/mmmurphy/projects/kronos/v1298tau_c/fits/starspots/outputs/contrasts/contrastfit_visit1_newera_fleck_R100_3comp_Tphotprior_v2rr_flatsamples.h5'
with h5py.File(path1, 'r') as f:
    fs1 = np.asarray(f['flatsamples'])

## -- visit 2 samples
path2 = '/home/mmmurphy/projects/kronos/v1298tau_bd/fits/spots/outputs/contrasts/contrastfit_visit2_newera_fleck8_R100_3comp_Tphotprior_v2rr_flatsamples.h5'
with h5py.File(path2, 'r') as f:
    fs2 = np.asarray(f['flatsamples'])
```

```python
fig = plt.figure(figsize=(5, 5))
labels = np.array([r'$T_{phot}$',r'$T_{umb}$',r'$T_{pen}$',r'$f_{umb}$'])

corner.corner(fs1, labels=labels, color='black', fig=fig, zorder=1, show_titles=False, labelpad=0.15, rasterized=True)
corner.corner(fs2, color='red', fig=fig, zorder=10, show_titles=False, rasterized=True)
handles = [
    Line2D([0], [0], color='black', lw=4, label='Visit 1'),
    Line2D([0], [0], color='red', lw=4, label='Visit 2'),
]
fig.legend(handles=handles, loc='upper right', bbox_to_anchor=(0.95, 0.95))
plt.savefig('./figures/Figure7.pdf', dpi=600, bbox_inches='tight')
plt.show()
```
